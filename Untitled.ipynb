```python
import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
```

```python
fashion_mnist,fashion_mnist_info=tfds.load('fashion_mnist',with_info=True,as_supervised=True)
```

```python
fashion_mnist_info
```

```python
train_data,test_data=fashion_mnist['train'],fashion_mnist['test']
```

```python
train_data
```

```python
num_validation_sample=int(0.1*fashion_mnist_info.splits['train'].num_examples)
```

```python
num_test_data=fashion_mnist_info.splits['test'].num_examples
```

```python
from sklearn.utils import shuffle
```

```python
buffer_size=10000
```

```python
shuffle_train_and_validation=train_data.shuffle(buffer_size)
```

```python
shuffled_validation=shuffle_train_and_validation.take(num_validation_sample)
```

```python
shuffled_train=shuffle_train_and_validation.skip(num_validation_sample)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
def Scaled(image,label):
    image = tf.cast(image, tf.float32)
    image/=255.
    return image,label
```

```python
scaled_data = fashion_mnist['train'].map(Scaled)
scaled_test_data= fashion_mnist['test'].map(Scaled)
```

```python
buffer_size = 1000

shuffle_train_and_validation = scaled_data.shuffle(buffer_size)
```

```python
shuffled_validation=shuffle_train_and_validation.take(num_validation_sample)
```

```python
shuffled_train=shuffle_train_and_validation.skip(num_validation_sample)
```

```python
batch_size=100
```

```python
train_datas=shuffled_train.batch(batch_size)
```

```python
validation_datas=shuffled_validation.batch(num_validation_sample)
```

```python
test_datas=scaled_test_data.batch(num_test_data)
```

```python
validation_input,validation_target=next(iter(validation_datas))
```

```python
input_size=784
hidden_layers_size=50
output_size=10

model=tf.keras.Sequential([
                           tf.keras.layers.Flatten(input_shape=(28, 28, 1)),
                           tf.keras.layers.Dense(hidden_layers_size, activation='relu'),
                           tf.keras.layers.Dense(hidden_layers_size, activation='relu'),
                           tf.keras.layers.Dense(output_size, activation='softmax')
])
```

```python
#custom_optimizer=tf.keras.optimizers.Adam(learning_rate=0.1)
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(train_datas,validation_data=[validation_input,validation_target],epochs=5)
```

```python
test_loss,test_accuracy=model.evaluate(test_datas)
```

```python
model.save('fashion_model.keras')
```
